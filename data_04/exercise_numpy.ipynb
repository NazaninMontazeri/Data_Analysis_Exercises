```python
import numpy as np
```

```python
x = np.zeros((3 , 2 , 4))

print(x)
```

```python
x = np.inf

print(x)
```

```python
y = np.array([[[1 , 2 , 3 , 4] , [5 , 6 , 7 , 8]]])

print(y.shape)
print(y.dtype)
```

```python
y = np.array([[[1 , 2 , 3 , 4] , [5 , 6 , 7 , 8]]] , dtype = np.uint8)

print(y.shape)
print(y.dtype)
```

```python
z = np.array([1 , 2 , np.nan])

print(z + 1)
```

```python
h = np.ones((3 , 4))

print(h)
print(h.dtype)
```

```python
k = np.full((2 , 3 , 4) , 7)

print(k)
```

```python
n = np.array([[1 , 2 , 3, 4 ] , [5 , 6 , 7 , 8]])
m = np.full(n.shape , 3.14)

print(n)
print(m)
```

```python
n = np.array([[1 , 2 , 3, 4 ] , [5 , 6 , 7 , 8]])
m = np.full_like(n , 3.14)

print(n)
print(m)
```
